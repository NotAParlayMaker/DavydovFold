# 2D-IR prediction
Compute an absorptive spectrum from a short trajectory.


```python
from julia.api import Julia
Julia(compiled_modules=False)
from julia import DavydovFoldon as df
traj, *_ = df.run_soliton("MAGLVF", tmax=2.0)
s = df.compute_2DIR_spectrum(traj)
import matplotlib.pyplot as plt
plt.imshow(s.signal[:, :, 0], origin="lower", cmap="RdBu_r")
plt.xlabel(r"$\omega_3$"); plt.ylabel(r"$\omega_1$");
```
