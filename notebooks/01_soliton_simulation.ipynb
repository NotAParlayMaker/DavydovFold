# Soliton simulation
Start Julia through PyJulia and plot the exciton kymograph.


```python
from julia.api import Julia
jl = Julia(compiled_modules=False)
from julia import DavydovFoldon as df
traj, theta, rho = df.run_soliton("MKFLILFN", tmax=2.0, dt=0.01)
import matplotlib.pyplot as plt
plt.imshow(traj.exciton_density, aspect="auto", origin="lower")
plt.xlabel("saved step"); plt.ylabel("residue"); plt.colorbar(label="exciton density");
```
